**Chapter-2: Data Cleaning** 

```python
food_df.head()
```

```python
food_df.dtypes
```

We are converting the item_price from a string object to float value

```python
price_in_float = []
for record in food_df['item_price']:
    price_in_float.append(float(record[1:]))
food_df['price_in_float'] = price_in_float
food_df.head()
```

```python
food_df.dtypes
```

This conversion can be done with chain of functions also

```python
food_df['item_price'] = food_df['item_price'].str.replace('$','').astype(float)
food_df.head()
```

```python
food_df.dtypes
```

```python
# Let's drop the price_in_float column bcoz it's not needed
food_df.drop('price_in_float',axis=1,inplace=True)
```

```python
food_df.describe(include='all')
```

```python
#Least 5 selling item based on revenue
food_df.groupby('item_name')['revenue'].sum().sort_values().head(5)
```

```python
for dept in depts:
    print("DEPARTMENT-->",dept)
    print()
    print(employee_data[employee_data['Department'] == dept])
    print()
```

```python
# Imputing salary one by one based on departments
# department - Ops
employee_data[(employee_data['Department'] == 'Ops') & (employee_data['Salary'].isnull())]
```

```python
employee_data.loc[(employee_data['Department'] == 'Ops') & (employee_data['Salary'].isnull()),'Salary'] = 60000.00
```

```python
#department - IT Tech
employee_data.loc[(employee_data['Department'] == 'IT Tech') & (employee_data['Salary'].isnull()) & (employee_data['Name'] == 'Roshan'),'Salary'] = 62000.00
```

```python
#department - Support
employee_data.loc[(employee_data['Department'] == 'Support') & (employee_data['Salary'].isnull()),'Salary'] = 61000.00
```

```python
employee_data.loc[(employee_data['Department'] == 'Support') & (employee_data['Age'].isnull()),'Age'] = 27
```

```python
employee_data
```

***LEVEL-3 - Try Imputing values with mean,median,mode***

for example, there is a null values in the salary column still left out. So I will try imputing it with the mean,median and mode

```python
#finding out mean
employee_data['Salary']
```

```python
# Let's find mean, median, mode
print("Mean:", employee_data['Salary'].mean())
print("Median:", employee_data['Salary'].median())
print("Mode:", employee_data['Salary'].mode())
```

*mode returns a series, that's why there is index 0 and then 60000.0 as output. So while using mode index the series and then use it.*

```python
# Imputing
employee_data['Salary'].fillna(round(employee_data['Salary'].mean(), 2), inplace=True)
employee_data
```
